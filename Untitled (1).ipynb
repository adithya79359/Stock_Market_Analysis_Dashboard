```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
#Download Stock Data
import yfinance as yf

stock = "AAPL"

df = yf.download(
    stock,
    start="2023-01-01",
    end="2024-01-01"
)
df.columns = df.columns.get_level_values(0)
print(df.columns)
```

```python
#Basic Information
print(df.info())
print(df.describe())
```

```python
# Closing Price Chart 
plt.figure(figsize=(12,5))

plt.plot(df["Close"])

plt.title("Apple Closing Price")
plt.xlabel("Date")
plt.ylabel("Price ($)")
plt.grid(True)

plt.show()
```

```python
#Moving Average 20 days
df["MA20"] = df["Close"].rolling(window=20).mean()

plt.figure(figsize=(12,5))

plt.plot(df["Close"],label="Close Price")
plt.plot(df["MA20"],label="20-Day MA")

plt.legend()

plt.title("Moving Average")

plt.show()
```

```python
#Moving Average 50 Days
df["MA50"] = df["Close"].rolling(50).mean()

plt.figure(figsize=(12,5))

plt.plot(df["Close"])
plt.plot(df["MA50"])

plt.title("50-Day Moving Average")

plt.show()
```

```python
#Daily Return
df["Daily Return"] = df["Close"].pct_change()
df.head()
```

```python
#Daily Return Graph
plt.figure(figsize=(12,5))

plt.plot(df["Daily Return"])

plt.title("Daily Return")

plt.grid(True)

plt.show()
```

```python
#Histogram of Returns
plt.figure(figsize=(8,5))
plt.hist(df["Daily Return"].dropna(),bins=40)
plt.title("Distribution of Daily Returns")
plt.show()
```

```python
#Highest Closing Price
print("Highest Close :",df["Close"].max())
```

```python
#Lowest Closing Price
print("Lowest Close :",df["Close"].min())
```

```python
#Candlestick Chart
fig = go.Figure(data=[go.Candlestick(
    x=df.index,
    open=df["Open"],
    high=df["High"],
    low=df["Low"],
    close=df["Close"]
)])

fig.update_layout(
    title="Apple Candlestick Chart",
    xaxis_title="Date",
    yaxis_title="Price"
)

fig.show()
```

```python
#Compare Multiple Companies
stocks = ["AAPL","MSFT","GOOGL","AMZN"]

compare = yf.download(
    stocks,
    start="2023-01-01",
    end="2024-01-01"
)["Close"]

compare.head()
```

```python
#Comparison Chart
plt.figure(figsize=(14,6))
for company in compare.columns:
    plt.plot(compare[company],label=company)
plt.legend()
plt.title("Stock Price Comparison")
plt.show()
```

```python
#Percentage Growth
returns = compare.pct_change()
returns.head()
```

```python
#Correlation
returns.corr()
```

```python
#Correlation Heatmap
plt.figure(figsize=(8,6))
plt.imshow(returns.corr())
plt.xticks(range(len(compare.columns)),compare.columns)
plt.yticks(range(len(compare.columns)),compare.columns)
plt.colorbar()
plt.show()
```

```python
#Highest Performing Stock
performance = ((compare.iloc[-1]-compare.iloc[0])/
               compare.iloc[0])*100

performance.sort_values(ascending=False)
```
